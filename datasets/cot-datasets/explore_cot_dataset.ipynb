```python
from datasets import load_dataset, Dataset
import json
```

```python
paths = [
        "outputs/easy/qwen3_32b_openai_gsm8k_outputs_evaluated.jsonl",
        "outputs/hard/qwen3_32b_LiveMathbench_evaluated.jsonl",
        "outputs/moderate/qwen3_32b_AI-MO_NuminaMath-CoT_cn_k12_outputs_evaluated.jsonl",
        "outputs/moderate/qwen3_32b_HuggingFaceH4_MATH-500_outputs_evaluated.jsonl",
    ]
```

```python
all_data = []
for path in paths:
    with open(path, "r") as f:
        for line in f:
            all_data.append(json.loads(line))
```

```python
data = Dataset.from_list(all_data)
```

```python
print(f"Dataset size: {len(data)}")
print(f"Dataset columns: {data.column_names}")
print(f"Dataset features: {data.features}")
print(f"Dataset shape: {data.shape}")
```

```python
correct_data = []

for i in range(len(data)):
    if data[i]["correctness_label"] == 1:
        correct_data.append(data[i])

print(f"Correct data size: {len(correct_data)}")
```
